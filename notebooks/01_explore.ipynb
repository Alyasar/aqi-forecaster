```python
import numpy as np
import os

df = pollution.copy()

# Remove rows with no timestamp
df = df.dropna(subset=["Timestamp"])

# The timestamps are already Indian time, just mislabeled as UTC.
# Read them, then drop the wrong label without shifting the clock.
df["Timestamp"] = (
    pd.to_datetime(df["Timestamp"], format="ISO8601", utc=True)
    .dt.tz_localize(None)
)

df["Timestamp"].head()
```

```python
rename_map = {
    "Timestamp": "time",
    "PM2.5 (µg/m³)": "pm25",
    "PM10 (µg/m³)": "pm10",
    "NO (µg/m³)": "no",
    "NO2 (µg/m³)": "no2",
    "NOx (ppb)": "nox",
    "NH3 (µg/m³)": "nh3",
    "SO2 (µg/m³)": "so2",
    "CO (mg/m³)": "co",
    "Ozone (µg/m³)": "o3",
    "Benzene (µg/m³)": "benzene",
    "Toluene (µg/m³)": "toluene",
}

df = df[list(rename_map.keys())].rename(columns=rename_map)
df.head()
```

```python
pollutants = [c for c in df.columns if c != "time"]

# Negative pollution is impossible, so it's sensor error
print("Negative values per column:")
print((df[pollutants] < 0).sum())
df[pollutants] = df[pollutants].mask(df[pollutants] < 0)

# Values above 1000 are beyond what CPCB analysers report
print("\nValues above 1000:")
print((df[pollutants] > 1000).sum())
df[pollutants] = df[pollutants].mask(df[pollutants] > 1000)

# Duplicate timestamps
print("\nDuplicate timestamps:", df["time"].duplicated().sum())
df = df.drop_duplicates(subset="time")
```

```python
hourly = df.set_index("time").resample("h").mean()

print("Hourly rows:", len(hourly))
print("\nMissing % per column:")
print((hourly.isna().mean() * 100).round(1))
```

```python
import matplotlib.pyplot as plt

hourly["pm25"].plot(figsize=(14, 4), title="PM2.5 at Shadipur, Delhi (hourly)")
plt.ylabel("µg/m³")
plt.show()
```

```python
os.makedirs("../data/processed", exist_ok=True)
hourly.to_csv("../data/processed/pollution_hourly.csv")
print("Saved!")
```

```python
w = weather.copy()
w["time"] = pd.to_datetime(w["time"])

w = w.rename(columns={
    "temperature_2m (°C)": "temp",
    "relative_humidity_2m (%)": "humidity",
    "dew_point_2m (°C)": "dew_point",
    "precipitation (mm)": "rain",
    "wind_speed_10m (km/h)": "wind_speed",
    "wind_direction_10m (°)": "wind_dir",
    "surface_pressure (hPa)": "pressure",
    "cloud_cover (%)": "cloud",
    "boundary_layer_height (m)": "blh",
}).set_index("time")

print("Weather rows:", len(w))
print("\nMissing % per column:")
print((w.isna().mean() * 100).round(1))
```

```python
data = hourly.join(w, how="inner")
print("Merged rows:", len(data))
data.head()
```

```python
print("Rows before:", len(data))
data = data.interpolate(method="time", limit=3)
data = data.dropna(subset=["pm25"])
print("Rows after:", len(data))
print("\nRemaining missing % per column:")
print((data.isna().mean() * 100).round(1))
```

```python
data.to_csv("../data/processed/merged.csv")
print("Saved merged data:", data.shape)
```

```python
blh_missing = data["blh"].isna()

print("Missing % by hour of day:")
print((blh_missing.groupby(data.index.hour).mean() * 100).round(0))

print("\nMissing % by month:")
print((blh_missing.groupby(data.index.to_period("M")).mean() * 100).round(0))
```

```python
blh_era5 = pd.read_csv("../data/raw/weather_blh_era5.csv", skiprows=3)
blh_era5["time"] = pd.to_datetime(blh_era5["time"])
blh_era5 = blh_era5.set_index("time")

# The file has one data column; take it whatever its exact name is
blh_era5 = blh_era5.iloc[:, 0].rename("blh")

print("ERA5 BLH missing %:", round(blh_era5.isna().mean() * 100, 1))

data["blh"] = blh_era5.reindex(data.index)

print("BLH missing in merged data now:", round(data["blh"].isna().mean() * 100, 1), "%")
print("\nMissing % by month:")
print((data["blh"].isna().groupby(data.index.to_period("M")).mean() * 100).round(0))
```

```python
data.to_csv("../data/processed/merged.csv")
print("Saved merged data:", data.shape)
```
